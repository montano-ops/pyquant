## 3. Conversions (log of the growth factor)

```python
import numpy as np
for s in [-0.50, 1.00]:
    print(f"simple {s:+.0%} -> log {np.log1p(s):+.3f}")
# -50% -> -0.693   |   +100% -> +0.693
```

Note the pair: losing half and doubling are the *same* log magnitude. Walk
down 50%, back up 100%: log-wealth returns to zero — compounding bookkeeping
made exact.

## 4. The six biases

1. **Look-ahead** — signal×same-day return; information from the future.
2. **Survivorship** — only today's alive firms in the sample; dead losers missing.
3. **Selection / cherry-picking** — best period/asset/spec reported.
4. **Data snooping / overfitting** — best of N searched configurations.
5. **Unrealistic costs/liquidity** — no spreads/impact/shorting reality.
6. **Multiple testing** — enough hypotheses tried, 5%-level "discoveries"
   appear by luck (the meta-bias that afflicts whole literatures).
