## 1. The chain

E[X] = Σx·p; Var(X) = E[X²]−(E[X])²; Var(aX+bY) = a²VarX + b²VarY +
2ab·Cov; portfolio: w₁²σ₁² + w₂²σ₂² + 2w₁w₂ρσ₁σ₂; SE(mean) = σ/√n; CLT:
sample means → N(μ, σ²/n) regardless of parent; t = (x̄−μ₀)/SE.

```python
import numpy as np
rng = np.random.default_rng(0)
x = rng.normal(0.001, 0.01, 10_000)
y = 0.5 * x + rng.normal(0, 0.008, 10_000)          # correlated pair
print(f"Var(aX+bY): sim {(0.3*x + 0.7*y).var():.8f} vs "
      f"formula {0.3**2*x.var() + 0.7**2*y.var() + 2*0.3*0.7*np.cov(x, y)[0,1]:.8f}")
print(f"SE(mean): sim over 2000 draws of n=100: "
      f"{np.array([rng.choice(x, 100).mean() for _ in range(2000)]).std():.6f} "
      f"vs σ/√100 = {x.std()/10:.6f}")
```

## 2. The two simulations

```python
# (a) SE(p̂)
for n in [63, 252, 1260]:
    ph = (rng.random((2000, n)) < 0.55).mean(axis=1)
    print(f"n={n}: formula {np.sqrt(0.55*0.45/n):.4f}, sim {ph.std():.4f}")

# (b) best of 500 fair strategies, 250 days
wins = (rng.random((500, 250)) < 0.5).mean(axis=1)
print(f"best of 500: {wins.max():.1%}")
```
