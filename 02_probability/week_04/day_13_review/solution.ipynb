## 2. Two computations

```python
import numpy as np
from scipy import stats

# (1) portfolio vol
v = 0.6**2 * 0.18**2 + 0.4**2 * 0.11**2 + 2*0.6*0.4*0.4*0.18*0.11
print(f"portfolio vol: {np.sqrt(v):.3f}")   # ≈ 0.132 -> 13.2%

# (2) 145/252 wins
phat = 145/252
z = (phat - 0.5) / np.sqrt(0.25/252)
print(f"z = {z:.2f}, exact one-sided p = {1 - stats.binom.cdf(144, 252, 0.5):.4f}")
```

Portfolio vol ≈ 13.2% (vs 15.2% if all in asset 1). z ≈ 2.31, exact
p ≈ 0.011 — evidence, but a 1%-per-quarter event among the hundreds of
strategies any shop evaluates is exactly what multiplicity manufactures
(preview of Harvey–Liu–Zhu, module 04).

## 3. The 63-day correlation experiment, cold

```python
rng = np.random.default_rng(0)
rhos = np.array([np.corrcoef(rng.normal(size=63), rng.normal(size=63))[0, 1]
                 for _ in range(2000)])
print(f"SD(ρ̂) = {rhos.std():.3f} (theory {1/np.sqrt(63):.3f})")
print(f"P(|ρ̂|>0.25) = {(abs(rhos) > 0.25).mean():.1%}")
```

SD ≈ 0.126; P(|ρ̂|>0.25) ≈ 5% (that is the 2-SE tail) — while |ρ̂|>0.126,
one SE, happens about a third of the time. If your blank-cell version
disagrees, the error is in the code, not the math.
