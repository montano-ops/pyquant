## 1. Rules verified

```python
import numpy as np
rng = np.random.default_rng(0)
r = rng.normal(0.0004, 0.011, 50_000)
A, B = r > 0.01, r > -0.01
lhs = (A | B).mean()
rhs = A.mean() + B.mean() - (A & B).mean()
print(f"union: {lhs:.5f} vs rule {rhs:.5f}")
print(f"conditional: {A[B].mean():.5f} vs rule {(A & B).mean() / B.mean():.5f}")
```

## 3. The decomposition

0.54×0.8% − 0.46×1.0% = 0.432% − 0.46% = **−0.028% per day** — negative!
Any re-slicing of the same data (by day-of-week, by regime, by signal)
must reproduce the same overall mean: deviations are *redistributions*,
never creations (until you condition on the future — which is why
conditioning direction is everything).

```python
print(0.54 * 0.008 - 0.46 * 0.010)
```

**Common mistake to log:** the win rate (54%) *feels* like the strategy
is good; expectancy says it loses. Win rate is not expectancy — ever.
